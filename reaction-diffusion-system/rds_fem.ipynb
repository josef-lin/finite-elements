# Reaction-Diffusion System
## Problem Setup
Consider a simply connected domain $\Omega$ with boundary $\partial \Omega$ and let $t \in I = [0,T]$. 
The model problem for the time-dependent PDE of the two=component reaction-diffusion system reads

$$
\frac{\partial u}{\partial t} - D_u \Delta u - f_1(u,v) = 0 \quad \text{for} \, x \in \Omega, \, t \in I, \\ 
\frac{\partial v}{\partial t} - D_v \Delta v - f_2(u,v) = 0 \quad \text{for} \, x \in \Omega, \, t \in I,  
\\ 
\nabla u \cdot n = \nabla v \cdot n = 0 \quad \text{for} \, x \in \partial \Omega, \, t \in I, \\
u(x(0), 0) = u_0, \quad v(x(0), 0) = v_0 \quad \text{for} \, x \in \Omega.
$$

The source functions $f_1, f_2$ may vary with space and time. We apply no flux boundary conditions. The initial conditions $u_0, v_0$ is a function of space only.
