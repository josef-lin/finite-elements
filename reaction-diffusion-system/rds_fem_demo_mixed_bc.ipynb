# Reaction-Diffusion System Mixed Boundary Condition Demo
## Problem Setup
Consider a simply connected domain $\Omega$ with boundary $\partial \Omega$ and let $t \in I = [0,T]$. 
The model problem for the time-dependent PDE of the two=component reaction-diffusion system reads

$$
\frac{\partial u}{\partial t} - D_u \Delta u - f_1(u,v) = 0 \, \text{in} \, x \in \Omega, \, t \in I, \\ 
\frac{\partial v}{\partial t} - D_v \Delta v - f_2(u,v) = 0 \, \text{in} \, x \in \Omega, \, t \in I,  
\\ 
\nabla u \cdot n = g_1, \, \nabla v \cdot n = g_2 \, \text{on} \, x \in \Lambda_N, \, t \in I, \\
u = u_D, \, v = v_D \, \text{on} \, x \in \Lambda_D, \, t \in I, \\
u(x(0), 0) = u_0, \, v(x(0), 0) = v_0 \, \text{in} \, x \in \Omega.
$$

The source functions $f_1, f_2$ may vary with space and time. We apply mixed boundary conditions, with $\Lambda_N$ and $\Lambda_D$ denoting the parts of the boundary $\partial \Omega$ where the Neumann and Dirichlet conditions apply. The initial conditions $u_0, v_0$ is a function of space only.

## The Variational Formulation

As we do for the heat equation, we discretize the time derivative by finite difference approximations. We wlet the superscript $n$ denote a quantity at time $t_n$, where $n$ is an integer counting time levels. E.g. $u^n$ denotes $u$ at the time step $n$. 

For time step $t_{n+1}$, we have the discretization for $u$
$$ 
\left(\frac{\partial u}{\partial t}\right)^{n+1} - D_u \Delta u^{n+1} - f_1^{n+1} = 0.
$$

Now, we highlight that the time derivative can be approximated by a difference quotient. We choose a simple backward difference so that we can apply the implicit Euler discretization.
$$
\left(\frac{\partial u}{\partial t}\right)^{n+1} \approx \frac{u^{n+1}-u^n}{\Delta t},
$$
where $\Delta t$ is the time discretization parameter, usually with $\Delta t = \frac{T_f}{N}$ where $T_f$ is the final time and $N$ is the number of discretization time steps. 

The substitution yields 
$$
\frac{u^{n+1}-u^n}{\Delta t} = D_u \Delta u^{n+1} + f_1^{n+1}.
$$

We can reorder the equation such that the left-hand side contains the terms with the unknowns $u^{n+1}$ and the right hand side containing only computed terms. 
The resulting equation is a sequence of stationary problems for $u^{n+1}$, assuming $u^n$ is known from solving the previous time step: 
$$
u^0 = u_0, \\
u^{n+1} - D_u \Delta t \Delta u^{n+1} = u^n + \Delta t f_1^{n+1}, \, n = 0, 1, 2, \dots 
$$

Finally, we can now use finite element method. 
We multiply by the test-function of $w \in \hat{W}$ and integrate second-order derivatives by parts. We demonstrate the integration by parts for the Laplacian. For notation simplicity, we introduc the symbol $u$ for $u^{n+1}$.

$$
-\int_\Omega \left( \nabla^2 u\right)wdx = \int_\Omega \nabla u \cdot \nabla w dx - \int_{\partial \Omega} \frac{\partial u}{\partial n} w ds. 
$$

Since we have the Dirichlet condition on $\Lambda_D$, we have $v=0$ on $\Lambda_D$ so that 
$$
- \int_{\partial \Omega} \frac{\partial u}{\partial n} w ds = - \int_{\Lambda_N} \frac{\partial u}{\partial n} w ds = \int_{\Lambda_N} g w ds,
$$
by appling the boundary condition on $\Lambda_N$. 

The resulting weak form reads:
$$
-\int_\Omega \left( \nabla^2 u\right)wdx = \int_\Omega \nabla u \cdot \nabla w dx + \int_{\Lambda_N} g w ds. 
$$

Finally, the resulting weak formulation in bilinear form and linear form:
$$ 
a_u(u,w) = L_{u, n+1}(w),
$$
where 
$$
a_u(u,w) = \int_\Omega \left( uw + D_u \Delta t \nabla u \cdot \nabla w \right) dx \\
L_{u, n+1}(w) = \int_\Omega \left(u^n + \Delta t f_1^{n+1} \right) \cdot w dx - D_u \Delta t \int_{\Lambda_N} g_1 w ds.
$$

We can do the same for the $v$ component so that we have the bilinear form and linear form: 
$$ 
a_v(v,w) = L_{v, n+1}(w),
$$
where 
$$
a_v(v,w) = \int_\Omega \left( vw + D_v \Delta t \nabla v \cdot \nabla w \right) dx \\
L_{v, n+1}(w) = \int_\Omega \left(v^n + \Delta t f_2^{n+1} \right) \cdot w dx - D_v \Delta t \int_{\Lambda_N} g_2 w ds.
$$

## A known analytical solution
We construct a test problem which makes it easy to determine if the calculations are correct. 

Choose the analytical solution to be 
$$
u = 1 + \alpha_1 x^2 + \alpha_2 y^2 + \alpha_3 t, \\
v = 1 + \beta_1 x^2 + \beta_2 y^2 + \beta_3 t.
$$

Then inserting into our original PDE, we find that 
$$
f_1(u,v) = - 2 D_u (\alpha_1 + \alpha_2)  + \alpha_3, \\
f_2(u,v) = - 2 D_v (\beta_1 + \beta_2) + \beta_3.
$$

Naturally, we choose the boundary values for the Dirichlet conditions
$$
u_d(x, y, t) = 1 + \alpha_1 x^2 + \alpha_2 y^2 + \alpha_3 t, \\
v_d(x, y, t) = 1 + \beta_1 x^2 + \beta_2 y^2 + \beta_3 t.
$$
and the initial values 
$$
u_0(x, y, t) = 1 + \alpha_1 x^2 + \alpha_2 y^2, \\
v_0(x, y, t) = 1 + \beta_1 x^2 + \beta_2 y^2.
$$

The Neumann boundary condition would be more complicated as naturally it depends on the domain. 

For simplicity sake, say the domain is the unit square $\Omega = [0,1] \times [0,1]$. 
Say the Dirichlet condition is applied on the left and right sides $x=0, 1$ and the Neumann condition is applied on the top and bottom sides $y=0, 1$. 
Then, we have 
$$
g_1 = 
\begin{cases} 
0, & \text{if } y = 0, \\
-2 D_u \alpha_2, & \text{if } y = 1,
\end{cases} \\
g_2 = \begin{cases} 
0, & \text{if } y = 0, \\
-2 D_v \beta_2, & \text{if } y = 1.
\end{cases}
$$
For ease of programming, we define $g_1, g_2$ as functions over the whole domain $\Omega$ such that $g_1, g_2$ takes correct values at the boundaries. One such extension is
$$
g_1 = -2 D_u \alpha_2 y, \\
g_2 = -2 D_v \beta_2 y.
$$

## Implementation

Imports

```python
from petsc4py import PETSc
from mpi4py import MPI 
import ufl 

from dolfinx import mesh, io, plot
from dolfinx.fem import (
    Constant,
    Function,
    functionspace,
    assemble_scalar,
    dirichletbc,
    form,
    locate_dofs_geometrical,
    extract_function_spaces,
)
from dolfinx.fem.petsc import (
    assemble_vector,
    assemble_matrix,
    create_vector,
    apply_lifting,
    set_bc,
)

import numpy as np 
import numpy.typing as npt

import matplotlib as mpl 
import pyvista 
```

Problem specific parameters

```python
t = 0.0 # start time 
T = 2.0 # final time 
num_steps = 20 # number of time steps 
dt = (T - t) / num_steps # time step size 
D_u = 1.0
alpha_1 = 1.0 
alpha_2 = 5.0 
alpha_3 = 3.0
D_v = 2.0
beta_1 = 2.0
beta_2 = -4.0
beta_3 = 1.0 
```

Define the mesh and appropriate function spaces. 

```python
nx, ny = 10, 10
domain = mesh.create_unit_square(MPI.COMM_WORLD, nx, ny, mesh.CellType.triangle) 
V = functionspace(domain, ("Lagrange", 1))
```

## Time-dependent output 
This is for visualizing in Paraview. 

```python
xdmf_u = io.XDMFFile(domain.comm, "u.xdmf", "w")
xdmf_u.write_mesh(domain)

xdmf_v = io.XDMFFile(domain.comm, "v.xdmf", "w")
xdmf_v.write_mesh(domain)
```

## Define the exact solution 

```python
class ExactSolution:
    def __init__(self, prm_1: float, prm_2: float, prm_3: float, t: float):
        self.prm_1 = prm_1
        self.prm_2 = prm_2
        self.prm_3 = prm_3
        self.t = t

    def __call__(self, x: npt.NDArray[np.floating]) -> npt.NDArray[np.floating]:
        return 1 + self.prm_1 * x[0] ** 2 + self.prm_2 * x[1] ** 2 + self.prm_3 * self.t


u_exact = ExactSolution(alpha_1, alpha_2, alpha_3, t)
v_exact = ExactSolution(beta_1, beta_2, beta_3, t)
```

## Defining the boundary condition 
To start, we begin with a Dirichlet boundary conditon. We interpolate the exact solution for our boundary. 

```python
def boundary_D(x):
    return np.logical_or(np.isclose(x[0], 0), np.isclose(x[0], 1))

dofs_D = locate_dofs_geometrical(V, boundary_D)

u_bc = Function(V)
u_bc.interpolate(u_exact)
bc_u = dirichletbc(u_bc, dofs_D)

v_bc = Function(V)
v_bc.interpolate(v_exact)
bc_v = dirichletbc(v_bc, dofs_D)
```

Next, we define the Neumann boundary condition. 
We first define $g_1 ,g_2$ using `UFL`'s `SpatialCoordinate`, then in turn create a boundary integration measure `ds`. We recall the test function $w$ is zero on the boundary integrals over the Dirichlet boundary disappears, so we can integrate $g v ds$ over the entire boundary. 

```python
x = ufl.SpatialCoordinate(domain) 
g1 = -2 * D_u * alpha_2 * x[1] 
g2 = -2 * D_v * beta_2 * x[1]
```

## Defining the variational formulation 

Having set $t=0$ in `u_exact`, we can reuse this variable to obtain $u_n$ for the initial time step. 

```python
u_n = Function(V)
u_n.interpolate(u_exact) 

v_n = Function(V)
v_n.interpolate(v_exact)
```

As $f_1, f_2$ are constants independent of $t$, we can define it as constants.

```python
f_1 = Constant(domain, -2 * D_u * (alpha_1 + alpha_2) + alpha_3) 
f_2 = Constant(domain, -2 * D_v * (beta_1 + beta_2) + beta_3)                
```

We can now create our variational formulation with the bilinear form `a_u`, `a_v` and the linear from `L_u`, `L_v`. 

We write the variational form on residual form, use `ufl.lhs()` and `ufl.rhs()` to extract the bilinear and linear forms. 

```python
u = ufl.TrialFunction(V)
v = ufl.TrialFunction(V)
w = ufl.TestFunction(V)

F_u = (
    u * w * ufl.dx
    + dt * D_u * ufl.dot(ufl.grad(u), ufl.grad(w)) * ufl.dx
    - (u_n + dt * f_1) * w * ufl.dx
    + dt * D_u * g1 * w * ufl.ds
)
a_u = form(ufl.lhs(F_u))
L_u = form(ufl.rhs(F_u))

F_v = (
    v * w * ufl.dx 
    + dt * D_v * ufl.dot(ufl.grad(v), ufl.grad(w)) * ufl.dx 
    - (v_n + dt * f_2) * w * ufl.dx 
    + dt * D_v * g2 * w * ufl.ds
)
a_v = form(ufl.lhs(F_v))
L_v = form(ufl.rhs(F_v))
```

## Create the matrix and vector for the linear problem
For computational efficiency, we create structures that can reuse data, such as matrix sparsity patterns. 
Here, the bilinear form `a` is independent of time, so only need to assemble the matrix once. 

```python
A_u = assemble_matrix(a_u, bcs = [bc_u])
A_u.assemble()
b_u = create_vector(extract_function_spaces(L_u))
uh = Function(V)
uh.name = "uh"

A_v = assemble_matrix(a_v, bcs = [bc_v])
A_v.assemble()
b_v = create_vector(extract_function_spaces(L_v))
vh = Function(V)
vh.name = "vh"
```

## Define a linear variational solver 
For the given problem we choose a direct solver and reuse the LU factorization at every time step. 

```python
solver_u = PETSc.KSP().create(domain.comm)
solver_u.setOperators(A_u)
solver_u.setType(PETSc.KSP.Type.PREONLY)
pc_u = solver_u.getPC()
pc_u.setType(PETSc.PC.Type.LU)

solver_v = PETSc.KSP().create(domain.comm)
solver_v.setOperators(A_v)
solver_v.setType(PETSc.KSP.Type.PREONLY)
pc_v = solver_v.getPC()
pc_v.setType(PETSc.PC.Type.LU)
```

## Visualization of time dependent problem using pyvista 
We use the DOLFINx plotting functionality, based on `pyvista` to plot every time step. We visualize a colorbar reflecting the minimal and maximal value of $u$ and $v$ at each time step. 

```python
grid_u = pyvista.UnstructuredGrid(*plot.vtk_mesh(V))

plotter_u = pyvista.Plotter()
plotter_u.open_gif("u_time.gif", fps=10)

grid_u.point_data["uh"] = uh.x.array
warped_u = grid_u.warp_by_scalar("uh", factor=1)

viridis = mpl.colormaps.get_cmap("viridis").resampled(25)
sargs = dict(
    title_font_size=25,
    label_font_size=20,
    fmt="%.2e",
    color="black",
    position_x=0.1,
    position_y=0.8,
    width=0.8,
    height=0.1,
)

renderer_u = plotter_u.add_mesh(
    warped_u,
    show_edges=True,
    lighting=False,
    cmap=viridis,
    scalar_bar_args=sargs,
    clim=[0, max(uh.x.array)],
)

grid_v = pyvista.UnstructuredGrid(*plot.vtk_mesh(V))

plotter_v = pyvista.Plotter()
plotter_v.open_gif("v_time.gif", fps=10)

grid_v.point_data["vh"] = vh.x.array
warped_v = grid_v.warp_by_scalar("vh", factor=1)

viridis = mpl.colormaps.get_cmap("viridis").resampled(25)
sargs = dict(
    title_font_size=25,
    label_font_size=20,
    fmt="%.2e",
    color="black",
    position_x=0.1,
    position_y=0.8,
    width=0.8,
    height=0.1,
)

renderer_v = plotter_v.add_mesh(
    warped_v,
    show_edges=True,
    lighting=False,
    cmap=viridis,
    scalar_bar_args=sargs,
    clim=[0, max(vh.x.array)],
)
```

## Solve the time-dependent problem 
We can now create our time-stepping loop. 
In the loop, we update the Dirichlet boundary condition by interpolating the updated expression `u_exact` and `v_exact` into `V`. 
We reassemble the right hand side vector `b` with the updates `u_n` and `v_n` then apply the boundary conditions to these vectors. 
We do this using the lifting operation. 

Then we solve the problem using PETSc nad update `u_n` and `v_n` with the data from `uh` and `vh`. 

```python
for n in range(num_steps):
    t += dt 

    # Update Diriclet boundary condition
    u_exact.t = t
    u_bc.interpolate(u_exact)
    v_exact.t = t
    v_bc.interpolate(v_exact)

    # Update the right hand side reusing the initial vector
    with b_u.localForm() as loc_b_u:
        loc_b_u.set(0)
    assemble_vector(b_u, L_u)

    with b_v.localForm() as loc_b_v:
        loc_b_v.set(0)
    assemble_vector(b_v, L_v)

    # Apply Dirichlet boundary condition to the vector
    apply_lifting(b_u, [a_u], [[bc_u]])
    b_u.ghostUpdate(addv=PETSc.InsertMode.ADD_VALUES, mode=PETSc.ScatterMode.REVERSE)
    set_bc(b_u, [bc_u])

    apply_lifting(b_v, [a_v], [[bc_v]])
    b_v.ghostUpdate(addv=PETSc.InsertMode.ADD_VALUES, mode=PETSc.ScatterMode.REVERSE)
    set_bc(b_v, [bc_v])

    # Solve linear problem
    solver_u.solve(b_u, uh.x.petsc_vec)
    uh.x.scatter_forward()
    solver_v.solve(b_v, vh.x.petsc_vec)
    vh.x.scatter_forward()

    # Update solution at previous time step (u_n)
    u_n.x.array[:] = uh.x.array
    v_n.x.array[:] = vh.x.array

    # Write solution to file
    xdmf_u.write_function(uh, t)
    xdmf_v.write_function(vh, t)

    # Update plot 
    new_warped_u = grid_u.warp_by_scalar("uh", factor=1)
    warped_u.points[:, :] = new_warped_u.points
    warped_u.point_data["uh"][:] = uh.x.array
    plotter_u.write_frame()
    new_warped_v = grid_v.warp_by_scalar("vh", factor=1)
    warped_v.points[:, :] = new_warped_v.points
    warped_v.point_data["vh"][:] = vh.x.array
    plotter_v.write_frame()
plotter_u.close()
plotter_v.close()
xdmf_u.close()
xdmf_v.close()
```

Finally, free the PETSc object to avoid memory leaks. 

```python
A_u.destroy()
b_u.destroy()
solver_u.destroy()

A_v.destroy()
b_v.destroy()
solver_v.destroy() 
```

## Verify the numerical solution
We compute the L2-error and the error at the mesh vertices for the last time step. 

```python
# Compute L2 error and error at nodes
V_ex = functionspace(domain, ("Lagrange", 2))

u_ex = Function(V_ex)
u_ex.interpolate(u_exact)

v_ex = Function(V_ex)
v_ex.interpolate(v_exact)

error_L2_u = np.sqrt(
    domain.comm.allreduce(
        assemble_scalar(form((uh - u_ex) ** 2 * ufl.dx)), op=MPI.SUM
    )
)
error_L2_v = np.sqrt(
    domain.comm.allreduce(
        assemble_scalar(form((vh - v_ex) ** 2 * ufl.dx)), op=MPI.SUM
    )
)
if domain.comm.rank == 0:
    print(f"L2-error u: {error_L2_u:.2e}")
    print(f"L2-error v: {error_L2_v:.2e}")

# Compute values at mesh vertices
error_max_u = domain.comm.allreduce(
    np.max(np.abs(uh.x.array - u_bc.x.array)), op=MPI.MAX
)
error_max_v = domain.comm.allreduce(
    np.max(np.abs(vh.x.array - v_bc.x.array)), op=MPI.MAX
)
if domain.comm.rank == 0:
    print(f"Error_max u: {error_max_u:.2e}")
    print(f"Error_max v: {error_max_v:.2e}")
```
