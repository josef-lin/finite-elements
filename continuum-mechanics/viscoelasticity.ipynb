# The equations of viscoelasticity

## The PDE Problem

The equations governing small viscoelastic deformations of a body $\Omega$ are given by

$$
\begin{aligned}
-\nabla \cdot \boldsymbol{\sigma}(\mathbf{u},\dot{\mathbf{u}})
&= \mathbf{f} && \text{in } \Omega, \\
\boldsymbol{\sigma}(\mathbf{u},\dot{\mathbf{u}})
&= \boldsymbol{\sigma}_{\mathrm{e}}(\mathbf{u})
+ \boldsymbol{\sigma}_{\mathrm{v}}(\dot{\mathbf{u}}), \\
\boldsymbol{\sigma}_{\mathrm{e}}(\mathbf{u})
&= \lambda \operatorname{tr}\bigl(\boldsymbol{\epsilon}(\mathbf{u})\bigr)\mathbf{I}
+ 2\mu \boldsymbol{\epsilon}(\mathbf{u}), \\
\boldsymbol{\sigma}_{\mathrm{v}}(\dot{\mathbf{u}})
&= \eta_\lambda \operatorname{tr}\bigl(\dot{\boldsymbol{\epsilon}}(\mathbf{u})\bigr)\mathbf{I}
+ 2\eta_\mu \dot{\boldsymbol{\epsilon}}(\mathbf{u}), \\
\boldsymbol{\epsilon}(\mathbf{u})
&= \frac{1}{2}\left(\nabla \mathbf{u} + (\nabla \mathbf{u})^T\right), \\
\dot{\boldsymbol{\epsilon}}(\mathbf{u})
&= \frac{\partial}{\partial t}\boldsymbol{\epsilon}(\mathbf{u})
= \frac{1}{2}\left(\nabla \dot{\mathbf{u}} + (\nabla \dot{\mathbf{u}})^T\right).
\end{aligned}
$$

where $\boldsymbol{\sigma}$ is the total stress tensor, $\boldsymbol{\sigma}_{\mathrm{e}}$ is the elastic stress tensor, $\boldsymbol{\sigma}_{\mathrm{v}}$ is the viscous stress tensor, $\mathbf{f}$ is the body force per unit volume, $\lambda$ and $\mu$ are the Lamé parameters, $\eta_\lambda$ and $\eta_\mu$ are the viscous material coefficients, $\mathbf{I}$ is the identity tensor, $\operatorname{tr}$ is the trace operator, $\boldsymbol{\epsilon}$ is the symmetric strain tensor, $\dot{\boldsymbol{\epsilon}}$ is the strain rate tensor, $\mathbf{u}$ is the displacement vector field, and $\dot{\mathbf{u}}$ is the displacement velocity.

The total stress is the sum of the elastic and viscous contributions:

$$
\boldsymbol{\sigma}
=
\boldsymbol{\sigma}_{\mathrm{elastic}}
+
\boldsymbol{\sigma}_{\mathrm{viscous}}.
$$

This formulation corresponds to the linear Kelvin--Voigt viscoelastic model, in which the elastic stress depends on strain and the viscous stress depends on strain rate.


## Weak formulation with IMEX time discretisation

To solve the viscoelastic problem in time, we introduce a uniform time discretization with time step $\Delta t$ and time levels $t_n=n\Delta t$. Let $\mathbf{u}^n$ denote the displacement field at time $t_n$. We approximate the displacement velocity using a backward difference:

$$
\dot{\mathbf{u}}^{n+1}
\approx
\frac{\mathbf{u}^{n+1}-\mathbf{u}^n}{\Delta t}.
$$

We adopt an implicit-explicit (IMEX) time-stepping scheme in which the elastic and viscous stress terms are treated implicitly, while the body force and prescribed boundary traction are treated explicitly. The variational formulation at time step $n+1$ is therefore

Find $\mathbf{u}^{n+1}\in V$ such that

$$
a(\mathbf{u}^{n+1},\mathbf{v})
=
L^n(\mathbf{v})
\qquad \forall \mathbf{v}\in\hat{V},
$$

where the bilinear form $a$ combines the elastic and viscous contributions:

$$
\begin{aligned}
a(\mathbf{u},\mathbf{v})
&=
a_{\mathrm{elastic}}(\mathbf{u},\mathbf{v})
+\frac{1}{\Delta t}a_{\mathrm{viscous}}(\mathbf{u},\mathbf{v}),
\end{aligned}
$$

and the right-hand-side functional is defined as

$$
\begin{aligned}
L^n(\mathbf{v})
&=
\int_\Omega \mathbf{f}^n\cdot\mathbf{v}\,d\mathbf{x}
+\int_{\partial\Omega_T}\mathbf{T}^n\cdot\mathbf{v}\,d\mathbf{s}\\
&\qquad
+\frac{1}{\Delta t}a_{\mathrm{viscous}}(\mathbf{u}^n,\mathbf{v}).
\end{aligned}
$$

The elastic and viscous bilinear forms are given by

$$
\begin{aligned}
a_{\mathrm{elastic}}(\mathbf{u},\mathbf{v})
&=
\int_\Omega
\boldsymbol{\sigma}_{\mathrm{elastic}}(\mathbf{u})
:\nabla\mathbf{v}\,d\mathbf{x},\\
a_{\mathrm{viscous}}(\mathbf{u},\mathbf{v})
&=
\int_\Omega
\boldsymbol{\sigma}_{\mathrm{viscous}}(\mathbf{u})
:\nabla\mathbf{v}\,d\mathbf{x},
\end{aligned}
$$

where

$$
\begin{aligned}
\boldsymbol{\sigma}_{\mathrm{elastic}}(\mathbf{u})
&=
\lambda(\nabla\cdot\mathbf{u})\mathbf{I}
+\mu\left(\nabla\mathbf{u}+(\nabla\mathbf{u})^T\right),\\
\boldsymbol{\sigma}_{\mathrm{viscous}}(\mathbf{u})
&=
\eta_\lambda(\nabla\cdot\mathbf{u})\mathbf{I}
+\eta_\mu\left(\nabla\mathbf{u}+(\nabla\mathbf{u})^T\right).
\end{aligned}
$$

At each time step, the unknown displacement $\mathbf{u}^{n+1}$ is obtained by solving the variational problem

$$
a(\mathbf{u}^{n+1},\mathbf{v})
=
L^n(\mathbf{v})
\qquad \forall \mathbf{v}\in\hat{V}.
$$

The bilinear form $a$ contains the implicit elastic and viscous contributions, while $L^n$ contains the explicitly evaluated body force and boundary traction, together with the known viscous contribution from the previous displacement.

The inner product of a symmtric tensor $A$ and an anti-symmetric tensor $B$ vanishes. So only the symmetric part of the product $\boldsymbol{\sigma} : \nabla \mathbf{v}$ survives, so we can replace $\nabla v$ with the symmetric gradient $\boldsymbol{\epsilon}(\mathbf{v})$ to get the variational form 
$$
a(\mathbf{u},\mathbf{v}) = \int_\Omega \boldsymbol{\sigma} : \boldsymbol{\epsilon}(v) d\mathbf{x}.
$$

## Example: A clamped viscoelastic beam under its own weight

We consider a clamped, box-shaped beam undergoing small viscoelastic deformations under gravity in three dimensions. The beam occupies the domain

$$
\Omega = (0,L)\times(0,W)\times(0,W),
$$

where $L$ is the length of the beam and $W$ is the width of its square cross section. We assume that the beam is homogeneous, isotropic, and governed by the linear Kelvin--Voigt viscoelastic model.

### Material properties and body force

The material is characterized by its density $\rho$, Lamé parameters $\lambda$ and $\mu$, and viscous coefficients $\eta_\lambda$ and $\eta_\mu$. The acceleration due to gravity is denoted by $g$.

The body force per unit volume is given by

$$
\mathbf{f} = (0,0,-\rho g).
$$

We assume that the beam is initially undeformed and at rest:

$$
\mathbf{u}(\mathbf{x},0)=\mathbf{0},
\qquad
\dot{\mathbf{u}}(\mathbf{x},0)=\mathbf{0}.
$$

### Boundary conditions

The beam is clamped at the end $x=0$, where the displacement is prescribed to be zero. We denote this boundary by

$$
\partial\Omega_D
=
\{0\}\times[0,W]\times[0,W],
$$

and impose the Dirichlet boundary condition

$$
\mathbf{u}=\mathbf{u}_D=\mathbf{0}
\qquad \text{on } \partial\Omega_D.
$$

The remaining boundary is traction free. Denoting it by $\partial\Omega_T=\partial\Omega\setminus\partial\Omega_D$, we impose

$$
\boldsymbol{\sigma}\cdot\mathbf{n}
=
\mathbf{T}=\mathbf{0}
\qquad \text{on } \partial\Omega_T,
$$

where $\mathbf{n}$ is the outward unit normal and $\boldsymbol{\sigma}$ is the total viscoelastic stress tensor.

```python
import pyvista 
from dolfinx import mesh, fem, plot, io, default_scalar_type
from dolfinx.fem.petsc import LinearProblem
from mpi4py import MPI
import ufl
import numpy as np

l = 1.0
w = 0.2

mu = 1.0
beta = 1.25
lambda_ = beta

eta_mu = 1.0
eta_lambda = 1.25

rho = 1.0
delta = w / l
gamma = 0.4 * delta**2
g = gamma

t = 0.0
Tf = 1.0
num_steps = 100
dt = Tf / num_steps
```

Then we create the mesh consisting of hexahedral elements. 

```python
domain = mesh.create_box(
    MPI.COMM_WORLD,
    [np.array([0, 0, 0]), np.array([L, W, W])],
    [20, 6, 6],
    cell_type=mesh.CellType.hexahedron,
)
V = fem.functionspace(domain, ("Lagrange", 1, (domain.geometry.dim,)))
```

Then we define the boundary conditions, the traction, and the integrating measures.

```python
def clamped_boundary(x):
    return np.isclose(x[0], 0)


fdim = domain.topology.dim - 1
boundary_facets = mesh.locate_entities_boundary(domain, fdim, clamped_boundary)

u_D = np.array([0, 0, 0], dtype=default_scalar_type)
bc = fem.dirichletbc(u_D, fem.locate_dofs_topological(V, fdim, boundary_facets), V)

T = fem.Constant(domain, default_scalar_type((0, 0, 0)))

dx = ufl.dx(domain=domain)
ds = ufl.Measure("ds", domain=domain)

def initial_condition(x, u0):
    values = np.zeros((3, x.shape[1]), dtype=default_scalar_type)
    return values + u0[:, None]
```

We now create our variational formulation using the above mathematical syntax. 

```python
def epsilon(u):
    return ufl.sym(
        ufl.grad(u)
    )  # Equivalent to 0.5*(ufl.nabla_grad(u) + ufl.nabla_grad(u).T)

def sigma_elastic(u):
    return lambda_ * ufl.nabla_div(u) * ufl.Identity(len(u)) + 2.0 * mu * epsilon(u)

def sigma_viscous(u):
    return eta_lambda * ufl.nabla_div(u) * ufl.Identity(len(u)) + 2.0 * eta_mu * epsilon(u)

# Initial displacement and velocity
u0 = np.array([0.0, 0.0, 0.0], dtype=default_scalar_type)
v0 = np.array([0.0, 0.0, 0.0], dtype=default_scalar_type)

u_n = fem.Function(V)
u_n.name = "u_n"
u_n.interpolate(lambda x: initial_condition(x, u0))

v_n = fem.Function(V)
v_n.name = "v_n"
v_n.interpolate(lambda x: initial_condition(x, v0))
```

```python
u = ufl.TrialFunction(V)
v = ufl.TestFunction(V)
f = fem.Constant(domain, default_scalar_type((0, 0, -rho * g)))

F = (
    ufl.inner(sigma_elastic(u), epsilon(v)) * dx
    + (1.0 / dt) * ufl.inner(sigma_viscous(u), epsilon(v)) * dx
    - ufl.dot(f, v) * dx
    - ufl.dot(T, v) * ds
    - (1.0 / dt) * ufl.inner(sigma_viscous(u_n), epsilon(v)) * dx
)

# Extract bilinear form and linear functional
a = fem.form(ufl.lhs(F))
L = fem.form(ufl.rhs(F))
```

```python
from dolfinx.fem.petsc import assemble_matrix, assemble_vector, create_vector

# Assemble system matrix
A = assemble_matrix(a, bcs=[bc])
A.assemble()

# Create right-hand-side vector
b = create_vector(fem.extract_function_spaces(L))
b.set(0.0)

# Initialize displacement solution from previous time step
uh = fem.Function(V)
uh.name = "uh"
uh.x.array[:] = u_n.x.array
uh.x.scatter_forward()
```

## Stress Computation 
Furthermore, we can compute various stress measures as soon as the displacement is computed. 

von Mises stress is defined by $\boldsymbol{\sigma}_m = \sqrt{\frac{3}{2}\mathbf{s}:\mathbf{s}}$ where $\mathbf{s}$ is the deviatoric stress tensor $\mathbf{s}(\mathbf{u}) = \boldsymbol{\sigma}(\mathbf{u}) - \frac{1}{3}\operatorname{tr}(\boldsymbol{\sigma}(\mathbf{u}))\mathbf{I}$.

The von Mises variable must be projected into an appropriate function space to visualize. 

```python
# Total viscoelastic stress at the current time step
u_dot = (uh - u_n) / dt

sigma_total = (
    sigma_elastic(uh)
    + sigma_viscous(u_dot)
)

# Deviatoric stress tensor
s = sigma_total - (1.0 / 3.0) * ufl.tr(sigma_total) * ufl.Identity(len(u))

# Von Mises stress
von_Mises = ufl.sqrt((3.0 / 2.0) * ufl.inner(s, s))

# Interpolate von Mises stress into a DG0 space
V_von_mises = fem.functionspace(domain, ("DG", 0))

stress_expr = fem.Expression(von_Mises, V_von_mises.element.interpolation_points)

stresses = fem.Function(V_von_mises)
stresses.name = "VonMises"
```

## Visualization

```python
vtx = io.VTXWriter(
    domain.comm,
    "viscoelasticity.bp",
    [uh, stresses],
    engine="BP4",
)
```

## Define a linear variational solver 
For the given problem we choose a direct solver and reuse the LU factorization at every time step. 

```python
from petsc4py import PETSc

solver = PETSc.KSP().create(domain.comm)
solver.setOperators(A)
solver.setType(PETSc.KSP.Type.PREONLY)
pc = solver.getPC()
pc.setType(PETSc.PC.Type.LU)
```

```python
for n in range(num_steps):
    t += dt

    # Update the right-hand side, reusing the existing vector
    with b.localForm() as loc_b:
        loc_b.set(0.0)

    assemble_vector(b, L)

    # Apply boundary conditions to the right-hand side
    # (For nonzero Dirichlet data, lifting and setting BC values are also needed.)
    fem.petsc.set_bc(b, [bc])

    # Solve the linear system
    solver.solve(b, uh.x.petsc_vec)
    uh.x.scatter_forward()

    # Update von Mises stress
    stress_expr = fem.Expression(von_Mises, V_von_mises.element.interpolation_points)
    stresses.interpolate(stress_expr)
    stresses.x.scatter_forward()

    # Write displacement and stress
    vtx.write(t)

    # Update displacement at the previous time step
    u_n.x.array[:] = uh.x.array
    u_n.x.scatter_forward()

vtx.close()
```

```python
A.destroy()
b.destroy()
solver.destroy()
```
